```python
%sql
USE CATALOG vertem;
USE SCHEMA workshop_dataprep;
```

```python
%sql
-- Tabela de "origem": simula o feed de CDC que chega do sistema
-- transacional (Debezium / Lakeflow Connect / Fivetran / etc).
-- E append-only de proposito: cada linha e um EVENTO, nao um estado.
CREATE OR REPLACE TABLE cdc_clientes_raw (
  cliente_id  INT       COMMENT 'Chave de negocio',
  nome        STRING,
  cidade      STRING,
  segmento    STRING    COMMENT 'Bronze / Prata / Ouro',
  operacao    STRING    COMMENT 'INSERT | UPDATE | DELETE',
  ts          TIMESTAMP COMMENT 'Momento do evento no sistema origem (sequencia)'
)
COMMENT 'Feed de CDC bruto de clientes - append only';
```
